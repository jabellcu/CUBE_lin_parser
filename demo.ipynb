## Modifying lines

Lines can be modified directly:

```python
print(ln)
```

```python
ln.CIRCULAR = 'T'
```

```python
# Nodes can be deleted:
del ln.nodes[1]
```

```python
# Modified:
ln.nodes[0].TF += 4
```

```python
# Or added:
ln.nodes.append(lin_parser.node('9999'))
```

```python
print(ln)
```

**Notice**: the original system is also modified

## ... lines!

We need some **nodes** to begind with:

```python
new_nodes = [lin_parser.node(n) for n in range(1, 15)]
```

```python
for n in new_nodes:
    if (n.ID % 3) == 1:
        n.TF = (n.ID / 2)**2
```

We can define some **lines** with those nodes:

```python
line_slow = lin_parser.line(new_nodes)
line_slow.NAME = 'Slow'
line_slow.HEADWAY = 15
line_slow.MODE = 1
line_slow.ONEWAY = 'T'
```

```python
line_slow
```

```python
line_fast = lin_parser.line(new_nodes)
line_fast.NAME = 'Fast'
line_fast.HEADWAY = 25
line_fast.MODE = 1
line_fast.ONEWAY = 'T'
```

```python
for n in line_fast.nodes:
    if not((n.ID % 3) == 1):
        n.stopping = False
```

```python
line_fast
```

## ... systems!

A **system** is just a list of comments and lines (*content*):

```python
new_system = lin_parser.system()  # empty!
```

```python
# Always use the content attribute. This preserves the order.
new_system.content += ["; Just a test system"]
new_system.content.extend([line_slow, line_fast])
```

```python
new_system
```

# Outputting systems

It is easy to produce **CUBE-compatible files**:

```python
system.save('example_rev.lin')  # comments and changes included
```

```python
new_system.save('example_new_system.lin')  # comments and changes included
```

**Nice**: User will be warned if there are lines with duplicated names. Output files will rename duplicated line names by default (but this can be de-activated).

## Lines with duplicated NAMEs

**Duplicated line names** can happen accidentally, when copying lines from system to system, for example. But there are warnigns in place to alert the user, and there are tools to manage the duplicates:

```python
system.lines['Y'].NAME = 'X'  # Stretching the system...
```

```python
system  # warning!
```

```python
print(system)  # print does not trigger any warning
```

```python
system.rename_dups()  # fix the mess!
```

```python
system.NAME_unique
```
