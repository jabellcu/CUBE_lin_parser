```python
import lin_parser
```

```python
sfilename = 'example.lin'
```

# Reading a file as a system

```python
system = lin_parser.system.read_file(sfilename)
```

```python
system  # Summary
```

```python
print(system)  # Details
```

**Systems** have comments and lines:

```python
system.comments
```

```python
system.lines
```

A system's **"lines" are a dictionary**, so we can loop through them easily:

```python
system.lines['X']
```

```python
system.lines.keys()
```

```python
for ln in system.lines.values():
    print(ln.NAME, ', ', ln.LONGNAME)
```

## Working with lines

```python
ln = system.lines['X']
```

```python
ln
```

```python
print(ln)
```

A **line** is just a sequence of nodes:

```python
ln.nodes
```

Some of the **nodes** are **stops**, the rest are there only to define the route:

```python
ln.stops
```

A **line** also has some attributes:

```python
ln.attrs
```

```python
ln.OPERATOR
```

```python
ln.attrs['HEADWAY[1]']
```

It is somethimes useful to use a line's **concatenation of stops**:

```python
ln.stop_seq
```

**Notice**: there are no intermediate nodes in the stop sequence, because they are not stops.

```python
system.lines_by_attr('MODE', 1)  # A list
```

```python
system.lines_query('''(ONEWAY == 'T') and ((CRUSHCAP) >= 70)''')
```

## Modifying lines

Lines can be modified directly:

```python
print(ln)
```

```python
ln.CIRCULAR = 'T'
```

```python
# Attributes with akward names can still be modified:
ln.__dict__['HEADWAY[1]'] *= 2  # hacky!
```

```python
# Nodes can be deleted:
del ln.nodes[1]
```

```python
# Modified:
ln.nodes[0].TF += 4
```

```python
# Or added:
ln.nodes.append(lin_parser.node('9999'))
```

```python
print(ln)
```

**Notice**: the original system is also modified

```python
print(system)
```
